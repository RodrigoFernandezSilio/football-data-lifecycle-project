```python
import os

import pandas as pd

from src.utils import get_project_root
```

```python
root = get_project_root()
```

```python
all_tables = pd.read_html('https://en.wikipedia.org/wiki/2026_FIFA_World_Cup', storage_options={"User-Agent": "Mozilla/5.0"})
```

```python
# Index of the first group-stage table within ALL_TABLES
FIRST_GROUP_TABLE_INDEX = 10

# Number of tables to skip to reach the next group table
GROUP_TABLE_INDEX_STEP = 7

# Total number of groups in the tournament (A–L)
TOTAL_GROUP_COUNT = 12
```

```python
group_tables = []

for i in range(
    FIRST_GROUP_TABLE_INDEX,
    FIRST_GROUP_TABLE_INDEX + GROUP_TABLE_INDEX_STEP * TOTAL_GROUP_COUNT,
    GROUP_TABLE_INDEX_STEP
):
    group_tables.append(all_tables[i])
```

```python
output_dir = f"{root}/data/raw"
os.makedirs(output_dir, exist_ok=True)

for i, table in enumerate(group_tables):
    letter = chr(ord("A") + i)
    table.to_csv(f"{root}/data/raw/fifa_world_cup_2026_group_{letter}.csv", index=False)
```

```python
clean_group_tables = []

for i, table in enumerate(group_tables):
    clean_table = table.copy()

    # Drop "Qualification" column
    clean_table = clean_table.drop(columns=["Qualification"])

    clean_table.columns.values[1] = "Team"

    clean_group_tables.append(clean_table)
```

```python
output_dir = f"{root}/data/clean"
os.makedirs(output_dir, exist_ok=True)

for i, table in enumerate(clean_group_tables):
    letter = chr(ord("A") + i)
    table.to_csv(f"{root}/data/clean/fifa_world_cup_2026_group_{letter}.csv", index=False)
```
