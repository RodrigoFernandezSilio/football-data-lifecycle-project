```python
import os

import pandas as pd

from src.utils import get_project_root
```

```python
root = get_project_root()
```

```python
def clean_match_scores_and_teams(df: pd.DataFrame) -> pd.DataFrame:
    '''Clean data by formatting scores, filtering valid matches, and splitting scores.'''

    df_clean = df.copy()

    # Clean score strings by removing spaces and non-numeric characters like 'a.e.t.'
    df_clean['score'] = df_clean['score'].str.replace(r'[^\d–]', '', regex=True)

    # Keep only rows with a valid 'number–number' score format
    df_clean = df_clean[df_clean['score'].str.contains(r'\d+–\d+', na=False, regex=True)].copy()

    # Remove leading and trailing whitespaces from team names
    df_clean['home'] = df_clean['home'].str.strip()
    df_clean['away'] = df_clean['away'].str.strip()

    # Split score into separate columns for home and away goals
    df_clean[['home_goals', 'away_goals']] = df_clean['score'].str.split('–', expand=True)

    # Remove the original score column as it is no longer needed
    df_clean.drop('score', axis=1, inplace=True)

    return df_clean
```

```python
df_fifa = pd.read_csv(f'{root}/data/raw/historical/fifa_world_cup_historical_data.csv')
```

```python
clean_df_fifa = clean_match_scores_and_teams(df_fifa)
```

```python
output_dir = f"{root}/data/clean/historical"
os.makedirs(output_dir, exist_ok=True)

clean_df_fifa.to_csv(f"{root}/data/clean/historical/fifa_world_cup_historical_data.csv", index=False)
```
