# Sandbox notebook for initial data source exploration


```python
import pandas as pd
```

```python
all_tables = pd.read_html('https://en.wikipedia.org/wiki/2022_FIFA_World_Cup', storage_options={"User-Agent": "Mozilla/5.0"})
```

```python
print(f"Number of tables: {len(all_tables)}")
```

```python
# Index of the first group-stage table within ALL_TABLES
FIRST_GROUP_TABLE_INDEX = 18

# Number of tables to skip to reach the next group table
GROUP_TABLE_INDEX_STEP = 7

# Total number of groups in the tournament (A–H)
TOTAL_GROUP_COUNT = 8
```

```python
# Confirm that all tables have the same schema

for i in range(
    FIRST_GROUP_TABLE_INDEX,
    FIRST_GROUP_TABLE_INDEX + GROUP_TABLE_INDEX_STEP * TOTAL_GROUP_COUNT,
    GROUP_TABLE_INDEX_STEP
):
    print(f"Table {i} columns:", all_tables[i].columns)
```
